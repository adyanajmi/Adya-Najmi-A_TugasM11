```python
# SOAL 3.1
def hitung_luas_persegi(sisi):
  hasil = sisi * sisi
  print(f"Luas persegi adalah: {hasil}")
  return hasil

def hitung_rata_rata(luas1, luas2, luas3):
  rata_rata = (luas1 + luas2 + luas3) / 3
  print(f"Rata-rata luas adalah: {rata_rata}")

hasil1 = hitung_luas_persegi(5)
hasil2 = hitung_luas_persegi(10)
hasil3 = hitung_luas_persegi(15)

hitung_rata_rata(hasil1, hasil2, hasil3)
```

```python
# SOAL 4
def salam(waktu, nama="Kawan"):
  print(f"Selamat {waktu}, {nama}!")

salam("Pagi", "Andi")
salam("Sore") #Hanya mengirimkan satu nilai untuk mengisi parameter waktu.
#Karena parameter kedua kosong, python mengambil nilai defaultnya (kawan) dan mencetak Selamat Sore, Kawan!.
```

```python
def konversi_nilai(nilai_angka):
  if  nilai_angka >= 86:
    return "A"
  elif nilai_angka >= 76 and nilai_angka < 86:
    return "AB"
  elif nilai_angka >= 66 and nilai_angka < 76:
    return "B"
  else:
    return "C"

nilai_input = int(input("Masukkan nilai: "))
indeks = konversi_nilai(nilai_input)
print(f"Nilai huruf Anda adalah: {indeks}")
```
